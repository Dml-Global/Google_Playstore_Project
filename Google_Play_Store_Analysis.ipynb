### Removal of Invalid/Malformed Record

During unique-value inspection, one record contained invalid values across several columns. 
The app "Life Made WI-Fi Touchscreen Photo Frame" had a Category value of `1.9`, 
a Rating of `19`, and several values shifted into incorrect columns.

Because the record was structurally corrupted and could not be reliably reconstructed, 
the entire row was removed rather than attempting to infer missing or displaced values.

```python
apps_clean.duplicated().sum()
```

```python
apps_clean.shape
```

```python
np.unique(apps_clean["Type"])
```

```python
apps_clean[
    apps_clean["Type"]
    .astype(str)
    .str.strip()
    .str.lower()
    .eq("nan")
].T
```

```python
for column in ["Reviews", "Installs", "Price"]:
    numeric_values = pd.to_numeric(apps_clean[column], errors="coerce")
    
    invalid_count = (
        numeric_values.isna() &
        apps_clean[column].astype(str).str.strip().ne("")
    ).sum()
    
    print(column, "- invalid non-numeric values:", invalid_count)
```

```python
rating_numeric = pd.to_numeric(
    apps_clean["Rating"],
    errors="coerce"
)

print("Blank ratings:", (apps_clean["Rating"] == "").sum())
print("Minimum valid rating:", rating_numeric.min())
print("Maximum valid rating:", rating_numeric.max())

invalid_ratings = (
    rating_numeric.isna() &
    apps_clean["Rating"].astype(str).str.strip().ne("")
).sum()

print("Invalid non-blank ratings:", invalid_ratings)
```

```python
updated_dates = pd.to_datetime(
    apps_clean["Last Updated"],
    format="%d-%m-%Y",
    errors="coerce"
)

print("Invalid dates:", updated_dates.isna().sum())
```

```python
for column in ["Type", "Current Ver", "Android Ver"]:
    
    blank_count = (
        apps_clean[column]
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )
    
    nan_text_count = (
        apps_clean[column]
        .astype(str)
        .str.strip()
        .str.lower()
        .eq("nan")
        .sum()
    )
    
    print(
        column,
        "| Blank:", blank_count,
        "| NaN text:", nan_text_count
    )
```

### Validation of Remaining Attributes

Following duplicate removal and deletion of the malformed record, the remaining columns were validated according to their expected characteristics.

- Reviews, Installs, and Price contained no invalid non-numeric values.
- Valid Rating values ranged from 1.0 to 5.0.
- 1,465 Rating values remained blank and were retained for subsequent null-value treatment.
- All Last Updated values were valid dates.
- Missing values remained in Type, Current Ver, and Android Ver and were retained for the spreadsheet-based null-value treatment required by the project brief.

```python
for column in apps_clean.columns:
    blank_count = (
        apps_clean[column]
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    nan_text_count = (
        apps_clean[column]
        .astype(str)
        .str.strip()
        .str.lower()
        .eq("nan")
        .sum()
    )

    if blank_count > 0 or nan_text_count > 0:
        print(
            column,
            "| Blank:", blank_count,
            "| NaN text:", nan_text_count
        )
```

### Removal of Duplicate Reviews

The raw Reviews dataset contained 64,295 records. A duplicate check identified 33,616 exact duplicate rows. These duplicates were removed using `drop_duplicates()`, leaving 30,679 unique records for further preprocessing.

```python
nan_reviews = (
    reviews_clean["Translated_Review"]
    .astype(str)
    .str.strip()
    .str.lower()
    .eq("nan")
    .sum()
)

blank_reviews = (
    reviews_clean["Translated_Review"]
    .astype(str)
    .str.strip()
    .eq("")
    .sum()
)

print("NaN translated reviews:", nan_reviews)
print("Blank translated reviews:", blank_reviews)
```

```python
unusable_reviews = (
    reviews_clean["Translated_Review"]
    .astype(str)
    .str.strip()
    .str.lower()
    .isin(["", "nan"])
)
```

```python
unusable_reviews.sum()
```

```python
reviews_clean = reviews_clean[
    ~unusable_reviews
].copy()
```

```python
reviews_clean.reset_index(drop=True, inplace=True)
```

```python
reviews_clean.shape
```

```python
for column in reviews_clean.columns:
    blank_count = (
        reviews_clean[column]
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    nan_count = (
        reviews_clean[column]
        .astype(str)
        .str.strip()
        .str.lower()
        .eq("nan")
        .sum()
    )

    print(
        column,
        "| Blank:", blank_count,
        "| NaN text:", nan_count
    )
```

```python
reviews_clean["Sentiment"].value_counts()
```

```python
reviews_clean["Sentiment_Polarity"] = pd.to_numeric(
    reviews_clean["Sentiment_Polarity"]
)

reviews_clean["Sentiment_Subjectivity"] = pd.to_numeric(
    reviews_clean["Sentiment_Subjectivity"]
)
```

```python
reviews_clean.dtypes
```
